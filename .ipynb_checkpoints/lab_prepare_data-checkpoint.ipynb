```python
import xarray as xr
import geopandas as gpd
import warnings

def clip_era5_memory_efficient(grib_path, shapefile_path, output_path, 
                                variables=None, buffer_deg=0.5, 
                                resample_daily=True, verbose=True):
    """
    Memory‑efficient clipping using bounding‑box subset.
    Opens the full GRIB once, selects variables by name, and subsets spatially.
    """
    # Load shapefile to get bounding box
    gdf = gpd.read_file(shapefile_path).to_crs("EPSG:4326")
    bbox = gdf.total_bounds
    lon_min = bbox[0] - buffer_deg
    lon_max = bbox[2] + buffer_deg
    lat_min = bbox[1] - buffer_deg
    lat_max = bbox[3] + buffer_deg
    
    if verbose:
        print(f"Bounding box: lon [{lon_min:.4f}, {lon_max:.4f}], lat [{lat_min:.4f}, {lat_max:.4f}]")
    
    # Open the full GRIB file with chunking (no variable filter)
    ds = xr.open_dataset(
        grib_path,
        engine='cfgrib',
        chunks={'time': 100, 'latitude': 100, 'longitude': 100},
        backend_kwargs={'filter_by_keys': {'typeOfLevel': 'surface'}}
    )
    
    # Detect dimension names (some files use 'lon'/'lat' instead)
    lon_dim = 'longitude' if 'longitude' in ds.dims else 'lon'
    lat_dim = 'latitude' if 'latitude' in ds.dims else 'lat'
    if verbose:
        print(f"Using dimensions: {lon_dim}, {lat_dim}")
    
    # If variables not specified, use all data_vars
    if variables is None:
        variables = list(ds.data_vars)
    else:
        # Only keep variables that exist
        available = list(ds.data_vars)
        variables = [v for v in variables if v in available]
        missing = set(variables) - set(available)
        if missing and verbose:
            print(f"Warning: Requested variables not found: {missing}")
    
    if not variables:
        raise ValueError(f"No requested variables found. Available: {available}")
    
    # Drop scalar 'step' coordinate if present
    if 'step' in ds.coords:
        ds = ds.drop_vars('step')
    
    # Spatial subset (lazy)
    ds_sub = ds.sel(**{lon_dim: slice(lon_min, lon_max), 
                       lat_dim: slice(lat_max, lat_min)})
    
    processed = {}
    for var in variables:
        if verbose:
            print(f"Processing {var}...")
        
        # Select the variable (still lazy)
        da = ds_sub[var]
        
        # Daily resampling
        if resample_daily:
            # Determine aggregation: sum for precipitation, mean for others
            if 'tp' in var or 'precip' in var or 'rain' in var:
                daily = da.resample(time='1D').sum()
                # Convert from metres to mm for precipitation
                daily = daily * 1000.0
                daily.attrs['units'] = 'mm'
            else:
                daily = da.resample(time='1D').mean()
            processed[var] = daily
        else:
            processed[var] = da
    
    # Close the original dataset to free resources
    ds.close()
    
    # Merge all processed variables
    output = xr.merge(processed.values())
    
    # Write to NetCDF with compression
    encoding = {var: {'zlib': True, 'complevel': 1, 'dtype': 'float32'} 
                for var in output.data_vars}
    output.to_netcdf(output_path, encoding=encoding, compute=True)
    if verbose:
        print(f"Clipped data saved to {output_path}")
    return output
```

```python
grib_file = "/mnt/monadworld/projects/phd/article_manuscripts/a_cdm_for_vector_of_vbds/data/datasets/climatic_timeseries_data/2020_data.grib"
shapefile = "/home/void/Documents/codes/monadvsim/prepared_data/dire_dawa/dire_dawa.shp"
output_nc = "/home/void/Documents/codes/monadvsim/prepared_data/dire_dawa/climate_t2m_2020.nc"

ds_clipped = clip_era5_memory_efficient(
    grib_path=grib_file,
    shapefile_path=shapefile,
    output_path=output_nc,
    buffer_deg=0.5,
    variables=['t2m'],   # only t2m exists
    resample_daily=True,
    verbose=True
)

# Verification
import netCDF4 as nc
with nc.Dataset(output_nc) as ds:
    print("Shape:", ds['t2m'].shape)
    print("First t2m values (Kelvin):", ds['t2m'][0,0,:5])
```
