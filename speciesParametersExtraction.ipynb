```python
import pandas as pd
import numpy as np
import os
import re


def extract_mean(value):
    """Convert '12.22 ± 2.85' or '12.22' to float mean."""
    if pd.isna(value):
        return np.nan
    if isinstance(value, (int, float)):
        return float(value)
    # Remove everything after '±' and strip spaces
    parts = str(value).split('±')[0].strip()
    # Also handle possible leading/trailing characters
    try:
        return float(parts)
    except:
        return np.nan

# ----------------------------------------------------------------------
# Standard output columns
# ----------------------------------------------------------------------
output_cols = [
    "trait.name", "T", "trait", "ref", "specie", "parasite",
    "food_supply", "sex", "notes"
]

master_dfs = []

def add_dataframe(df):
    for col in output_cols:
        if col not in df.columns:
            df[col] = ""
    master_dfs.append(df[output_cols])

   
villena_data["food_supply"] = ""
villena_data["sex"] = ""
villena_data["notes"] = ""
add_dataframe(villena_data)
print("Loaded Villena data")

dev = nokubo_data_1.copy()
dev["trait.name"] = "mdr"
dev["trait"] = 1.0 / nokubo_data_1["Larval Period (Days) / N"]
dev["ref"] = "Moghadam2023"
dev["specie"] = "An. stephensi"
dev["parasite"] = ""
dev["food_supply"] = dev["Food Supply"]
dev["sex"] = ""
dev["notes"] = "larval development rate (1/days)"
if "Temp. (°C)" in dev.columns:
    dev["T"] = dev["Temp. (°C)"]
else:
    dev["T"] = ""
add_dataframe(dev[output_cols])
    
surv = nokubo_data_1.copy()
surv["trait.name"] = "larva_survival"
surv["trait"] = 1.0 - nokubo_data_1["Larval Mortality / %"] / 100.0
surv["ref"] = "Moghadam2023"
surv["specie"] = "An. stephensi"
surv["food_supply"] = nokubo_data_1["Food Supply"]
surv["notes"] = "larval survival (1 - mortality)"
surv["parasite"] = ""
surv["sex"] = ""
if "Temp. (°C)" in surv.columns:
    surv["T"] = surv["Temp. (°C)"]
else:
    surv["T"] = ""
add_dataframe(surv[output_cols])
print("Loaded Nokubo Table1")


nokubo_data_3["trait.name"] = "efd"
nokubo_data_3["trait"] = nokubo_data_3["No. of Eggs Oviposited / Mean ± SD"]
nokubo_data_3["ref"] = "Moghadam2023"
nokubo_data_3["specie"] = "An. stephensi"
nokubo_data_3["food_supply"] = "high"
nokubo_data_3["sex"] = "female"
nokubo_data_3["parasite"] = ""
nokubo_data_3["notes"] = "eggs oviposited"
if "Temp. (°C)" in nokubo_data_3.columns:
    nokubo_data_3["T"] = nokubo_data_3["Temp. (°C)"]
else:
    nokubo_data_3["T"] = ""
add_dataframe(nokubo_data_3[output_cols])
print("Loaded Nokubo Table3")


nokubo_data_4["trait.name"] = "bc.succ"
nokubo_data_4["trait"] = nokubo_data_4["One Blood Meal / Oocysts / (%)"] / 100.0
nokubo_data_4["ref"] = "Moghadam2023"
nokubo_data_4["specie"] = "An. stephensi"
nokubo_data_4["parasite"] = "P. berghei"
nokubo_data_4["food_supply"] = nokubo_data_4["Food Supply"]
nokubo_data_4["sex"] = "female"
nokubo_data_4["notes"] = "proportion with oocysts"
if "Temp. (°C)" in nokubo_data_4.columns:
    nokubo_data_4["T"] = nokubo_data_4["Temp. (°C)"]
else:
    nokubo_data_4["T"] = ""
add_dataframe(nokubo_data_4[output_cols])
print("Loaded Nokubo Table4")

# ----------------------------------------------------------------------
# 3. Reisen & Mahmood (1980) – tables with columns: ATTRIBUTE, An. culicifacies, An. stephensi
#    The delimiter is likely whitespace (space/tab). We'll use regex '\s+'.
# ----------------------------------------------------------------------
def load_reisen_table(path, table_name):
    # Try reading with whitespace delimiter
    df = pd.read_csv(path, sep=r'|', engine='python')
    # If that fails, try with no explicit separator (auto-detect)
    if df.shape[1] < 2:
        df = pd.read_csv(path, sep=None, engine='python')
    # Ensure we have the expected columns
    if 'ATTRIBUTE' not in df.columns or 'An. stephensi' not in df.columns:
        print(f"Warning: Reisen {table_name} does not have expected columns. Found: {df.columns.tolist()}")
        return
    
    mapping = {
        "e_1 Female": ("adult_life_expectancy", "female", "life expectancy at emergence (days)"),
        "e_1 Male": ("adult_life_expectancy", "male", "life expectancy at emergence (days)"),
        "s Female": ("adult_survival_daily", "female", "daily survival probability (type II)"),
        "s Male": ("adult_survival_daily", "male", "daily survival probability (type II)"),
        "R_0": ("net_reproductive_rate", "both", "net reproductive rate (female offspring per female)"),
        # Add more attributes as they appear in your file
    }
    rows = []
    for _, row in df.iterrows():
        attr = str(row["ATTRIBUTE"]).strip()
        if attr in mapping:
            trait_name, sex, note = mapping[attr]
            val_str = row["An. stephensi"]
            mean = extract_mean(val_str)
            if not np.isnan(mean):
                rows.append({
                    "trait.name": trait_name,
                    "T": 28,   # temperature not reported, assume ~28°C? leave blank
                    "trait": mean,
                    "ref": "Reisen1980",
                    "specie": "An. stephensi",
                    "parasite": "",
                    "food_supply": "",
                    "sex": sex,
                    "notes": note,
                })
    if rows:
        df_reisen = pd.DataFrame(rows)
        add_dataframe(df_reisen)
        print(f"Loaded Reisen {table_name} with {len(rows)} rows")
    else:
        print(f"No matching attributes in Reisen {table_name}")

# Adjust paths to your actual Reisen files
if os.path.exists("/mnt/monadworld/projects/phd/data/1980.WK.HLT/table1.csv"):
    load_reisen_table("/mnt/monadworld/projects/phd/data/1980.WK.HLT/table1.csv", "Table1")
if os.path.exists("/mnt/monadworld/projects/phd/data/1980.WK.HLT/table2.csv"):
    load_reisen_table("/mnt/monadworld/projects/phd/data/1980.WK.HLT/table2.csv", "Table2")

# ----------------------------------------------------------------------
# 4. Merge and save
# ----------------------------------------------------------------------
if master_dfs:
    final_df = pd.concat(master_dfs, ignore_index=True)
    final_df.to_csv("aggregated_traits.csv", index=False)
    print(f"\n✅ Merged file saved: aggregated_traits.csv ({len(final_df)} rows)")
    print("\nFirst few rows:")
    print(final_df.head(10))
else:
    print("No data loaded – check your file paths and column names.")
```

#### Model fitting to extract Species simulation parameters - Option 1

```python
import pandas as pd
import numpy as np
from scipy.optimize import curve_fit
import yaml

# ------------------------------------------------------------
# Helper: convert numpy scalars to Python floats
# ------------------------------------------------------------
def to_float(value):
    if hasattr(value, 'item'):  # numpy scalar
        return float(value.item())
    if isinstance(value, (list, tuple)):
        return [to_float(v) for v in value]
    if isinstance(value, dict):
        return {k: to_float(v) for k, v in value.items()}
    return value

# ------------------------------------------------------------
# 1. Load aggregated data
# ------------------------------------------------------------
df = pd.read_csv("aggregated_traits.csv")
df = df[df["specie"] == "An. stephensi"]

def get_trait_data(trait_name):
    sub = df[df["trait.name"] == trait_name].copy()
    sub = sub.dropna(subset=["T", "trait"])
    return sub["T"].values.astype(float), sub["trait"].values.astype(float)

# ------------------------------------------------------------
# 2. Define functions
# ------------------------------------------------------------
def briere(T, a, T0, Tm, m):
    val = a * T * (T - T0) * np.power(np.maximum(Tm - T, 0), 1.0/m)
    return np.maximum(val, 0)

def exp_dev(T, rho, k, Delta, lam):
    exp_arg = rho * k - (k - T) / Delta
    exp_arg = np.clip(exp_arg, -700, 700)
    val = np.exp(rho * T - np.exp(exp_arg)) + lam
    return np.maximum(val, 0)

def exp_quad(T, b1, b2, b3):
    return np.exp(b1 + b2 * T + b3 * T**2)

def gaussian_on_log(T, rmax, Topt, c):
    return np.exp(rmax + c * (Topt - T)**2)

# ------------------------------------------------------------
# 3. Larval development (Brière)
# ------------------------------------------------------------
T_mdr, y_mdr = get_trait_data("mdr")
if len(T_mdr) > 0:
    p0 = [1e-5, 10, 40, 2]
    bounds = ([0, 0, 20, 1], [1e-3, 20, 45, 5])
    popt, _ = curve_fit(briere, T_mdr, y_mdr, p0=p0, bounds=bounds)
    larva_dev = {"a": popt[0], "Tmin": popt[1], "Tmax": popt[2], "m": popt[3]}
else:
    larva_dev = {"a": 3.965e-5, "Tmin": 11.15, "Tmax": 41.41, "m": 1.425}

# ------------------------------------------------------------
# 4. Fecundity (Gaussian‑on‑log)
# ------------------------------------------------------------
T_efd, y_efd = get_trait_data("efd")
if len(T_efd) > 0:
    p0 = [2.0, 28, -0.01]
    bounds = ([0, 15, -0.1], [5, 40, 0])
    popt, _ = curve_fit(gaussian_on_log, T_efd, y_efd, p0=p0, bounds=bounds)
    fecundity = {"rmax": popt[0], "Topt": popt[1], "c": popt[2]}
else:
    fecundity = {"rmax": 2.6749, "Topt": 29.393, "c": -0.004114}

# ------------------------------------------------------------
# 5. Adult mortality (exp‑quadratic)
# ------------------------------------------------------------
T_mu, y_mu = get_trait_data("mu")
if len(T_mu) > 0:
    p0 = [0.0, 0.0, 0.0]
    bounds = ([-10, -1, -0.1], [10, 1, 0.1])
    popt, _ = curve_fit(exp_quad, T_mu, y_mu, p0=p0, bounds=bounds)
    adult_mort = {"b1": popt[0], "b2": popt[1], "b3": popt[2]}
    adult_mort_const = np.mean(y_mu)
else:
    adult_mort = {"b1": -1.4775, "b2": -0.1377, "b3": 0.003908}
    adult_mort_const = 0.1198

# ------------------------------------------------------------
# 6. Larval mortality from survival data
# ------------------------------------------------------------
T_surv, y_surv = get_trait_data("larva_survival")
if len(T_surv) > 0 and len(T_mdr) > 0:
    surv_df = pd.DataFrame({"T": T_surv, "survival": y_surv})
    dev_df = pd.DataFrame({"T": T_mdr, "rate": y_mdr})
    merged = pd.merge(surv_df, dev_df, on="T")
    if len(merged) > 0:
        merged["mu_daily"] = -np.log(merged["survival"]) * merged["rate"]
        T_mu_l = merged["T"].values
        y_mu_l = merged["mu_daily"].values
        p0 = [0.0, 0.0, 0.0]
        bounds = ([-10, -1, -0.1], [10, 1, 0.1])
        popt, _ = curve_fit(exp_quad, T_mu_l, y_mu_l, p0=p0, bounds=bounds)
        larva_mort = {"b1": popt[0], "b2": popt[1], "b3": popt[2]}
    else:
        larva_mort = {"b1": 3.572876, "b2": -0.323474, "b3": 0.004941}
else:
    larva_mort = {"b1": 3.572876, "b2": -0.323474, "b3": 0.004941}

# ------------------------------------------------------------
# 7. Egg and pupa mortality (defaults)
# ------------------------------------------------------------
egg_mort = {"b1": 3.572876, "b2": -0.323474, "b3": 0.004941}
pupa_mort = {"b1": 5.882576, "b2": -0.578528, "b3": 0.009458}

# ------------------------------------------------------------
# 8. Egg and pupa development – fit exp_dev to Mordecai quadratics
# ------------------------------------------------------------
def mordecai_quad(T, a, b, c):
    return np.maximum(0, a*T**2 + b*T + c)

def fit_exp_dev(quad_coeff, T_range):
    T_vals = np.linspace(T_range[0], T_range[1], 50)
    y_vals = mordecai_quad(T_vals, **quad_coeff)
    mask = y_vals > 0
    T_fit = T_vals[mask]
    y_fit = y_vals[mask]
    p0 = [0.01, 40, 1, -1]
    bounds = ([0, 20, 0.5, -5], [0.1, 45, 3, 0])
    popt, _ = curve_fit(exp_dev, T_fit, y_fit, p0=p0, bounds=bounds)
    return {"rho": popt[0], "k": popt[1], "Delta": popt[2], "lambda": popt[3]}

egg_quad = {"a": -0.00098, "b": 0.0502, "c": -0.348}
pupa_quad = {"a": -0.00074, "b": 0.0398, "c": -0.291}
egg_dev = fit_exp_dev(egg_quad, (15, 35))
pupa_dev = fit_exp_dev(pupa_quad, (15, 35))

# ------------------------------------------------------------
# 9. Build YAML with float conversion
# ------------------------------------------------------------
species_dict = {
    "egg_dev_rho": egg_dev["rho"],
    "egg_dev_k": egg_dev["k"],
    "egg_dev_Delta": egg_dev["Delta"],
    "egg_dev_lambda": egg_dev["lambda"],
    "larva_dev_a": larva_dev["a"],
    "larva_dev_Tmin": larva_dev["Tmin"],
    "larva_dev_Tmax": larva_dev["Tmax"],
    "larva_dev_m": larva_dev["m"],
    "pupa_dev_rho": pupa_dev["rho"],
    "pupa_dev_k": pupa_dev["k"],
    "pupa_dev_Delta": pupa_dev["Delta"],
    "pupa_dev_lambda": pupa_dev["lambda"],
    "egg_mort_b1": egg_mort["b1"],
    "egg_mort_b2": egg_mort["b2"],
    "egg_mort_b3": egg_mort["b3"],
    "larva_mort_b1": larva_mort["b1"],
    "larva_mort_b2": larva_mort["b2"],
    "larva_mort_b3": larva_mort["b3"],
    "pupa_mort_b1": pupa_mort["b1"],
    "pupa_mort_b2": pupa_mort["b2"],
    "pupa_mort_b3": pupa_mort["b3"],
    "fecundity_rmax": fecundity["rmax"],
    "fecundity_Topt": fecundity["Topt"],
    "fecundity_c": fecundity["c"],
    "adult_mort_b1": adult_mort["b1"],
    "adult_mort_b2": adult_mort["b2"],
    "adult_mort_b3": adult_mort["b3"],
    "adult_mortality_per_day": adult_mort_const,
    "sex_ratio": 0.5,
}

# Convert all numpy scalars to Python floats
species_dict_clean = to_float(species_dict)

species_config = {"species": species_dict_clean}

with open("species_parameters_option1.yaml", "w") as f:
    yaml.dump(species_config, f, sort_keys=False, indent=2, default_flow_style=False)

print("✅ Option 1 YAML saved as species_parameters_option1.yaml")
```

#### Model Fitting to extract species simulation parameter - Option 2 Mordecai

```python
import pandas as pd
import numpy as np
from scipy.optimize import curve_fit
import yaml

# ------------------------------------------------------------
# Helper
# ------------------------------------------------------------
def to_float(value):
    if hasattr(value, 'item'):
        return float(value.item())
    if isinstance(value, (list, tuple)):
        return [to_float(v) for v in value]
    if isinstance(value, dict):
        return {k: to_float(v) for k, v in value.items()}
    return value

# ------------------------------------------------------------
# 1. Load aggregated data
# ------------------------------------------------------------
df = pd.read_csv("aggregated_traits.csv")
df = df[df["specie"] == "An. stephensi"]

def get_trait_data(trait_name):
    sub = df[df["trait.name"] == trait_name].copy()
    sub = sub.dropna(subset=["T", "trait"])
    return sub["T"].values.astype(float), sub["trait"].values.astype(float)

# ------------------------------------------------------------
# 2. Define functions
# ------------------------------------------------------------
def quadratic(T, a, b, c):
    return np.maximum(0, a*T**2 + b*T + c)

def gaussian_survival(T, amp, mean, sigma):
    return amp * np.exp(-0.5 * ((T - mean) / sigma)**2)

def fecundity_asymmetric(T, a, b, Tmax, c):
    term1 = a * np.exp(b * T)
    term2 = np.exp(b * Tmax - ((Tmax - T) / c)**2)
    return np.maximum(0, term1 - term2)

# ------------------------------------------------------------
# 3. Larval development (quadratic)
# ------------------------------------------------------------
T_mdr, y_mdr = get_trait_data("mdr")
if len(T_mdr) > 0:
    p0 = [0.0001, -0.01, 0.2]
    popt, _ = curve_fit(quadratic, T_mdr, y_mdr, p0=p0)
    larva_dev = {"a": popt[0], "b": popt[1], "c": popt[2]}
else:
    larva_dev = {"a": -0.00082, "b": 0.0425, "c": -0.305}

# ------------------------------------------------------------
# 4. Larval survival (Gaussian)
# ------------------------------------------------------------
T_surv, y_surv = get_trait_data("larva_survival")
if len(T_surv) > 0:
    p0 = [0.9, 27, 5]
    bounds = ([0, 15, 1], [1, 40, 15])
    popt, _ = curve_fit(gaussian_survival, T_surv, y_surv, p0=p0, bounds=bounds)
    larva_surv = {"amp": popt[0], "mean": popt[1], "sigma": popt[2]}
else:
    larva_surv = {"amp": 0.88, "mean": 26.8, "sigma": 3.8}

# ------------------------------------------------------------
# 5. Fecundity (asymmetric)
# ------------------------------------------------------------
T_efd, y_efd = get_trait_data("efd")
if len(T_efd) > 0:
    p0 = [-0.01, 0.5, 32, 3]
    bounds = ([-1, 0, 25, 1], [0, 2, 40, 10])
    popt, _ = curve_fit(fecundity_asymmetric, T_efd, y_efd, p0=p0, bounds=bounds)
    fecundity = {"a": popt[0], "b": popt[1], "Tmax": popt[2], "c": popt[3]}
else:
    fecundity = {"a": -0.0099, "b": 0.567, "Tmax": 32.0, "c": 2.97}

# ------------------------------------------------------------
# 6. Adult mortality (quadratic)
# ------------------------------------------------------------
T_mu, y_mu = get_trait_data("mu")
if len(T_mu) > 0:
    p0 = [0.0001, -0.001, 0.05]
    bounds = ([0, -0.1, 0], [0.01, 0, 1])
    popt, _ = curve_fit(quadratic, T_mu, y_mu, p0=p0, bounds=bounds)
    adult_mort = {"a": popt[0], "b": popt[1], "c": popt[2]}
else:
    adult_mort = {"a": 0.00012, "b": -0.0018, "c": 0.055}

# ------------------------------------------------------------
# 7. Egg and pupa (Mordecai defaults)
# ------------------------------------------------------------
egg_dev = {"a": -0.00098, "b": 0.0502, "c": -0.348}
pupa_dev = {"a": -0.00074, "b": 0.0398, "c": -0.291}
egg_surv = {"amp": 0.94, "mean": 26.2, "sigma": 3.5}
pupa_surv = {"amp": 0.91, "mean": 26.5, "sigma": 3.6}

# ------------------------------------------------------------
# 8. Build YAML with float conversion
# ------------------------------------------------------------
species_dict = {
    "egg_dev_a": egg_dev["a"],
    "egg_dev_b": egg_dev["b"],
    "egg_dev_c": egg_dev["c"],
    "larva_dev_a": larva_dev["a"],
    "larva_dev_b": larva_dev["b"],
    "larva_dev_c": larva_dev["c"],
    "pupa_dev_a": pupa_dev["a"],
    "pupa_dev_b": pupa_dev["b"],
    "pupa_dev_c": pupa_dev["c"],
    "egg_survival_amp": egg_surv["amp"],
    "egg_survival_mean": egg_surv["mean"],
    "egg_survival_sigma": egg_surv["sigma"],
    "larva_survival_amp": larva_surv["amp"],
    "larva_survival_mean": larva_surv["mean"],
    "larva_survival_sigma": larva_surv["sigma"],
    "pupa_survival_amp": pupa_surv["amp"],
    "pupa_survival_mean": pupa_surv["mean"],
    "pupa_survival_sigma": pupa_surv["sigma"],
    "fecundity_a": fecundity["a"],
    "fecundity_b": fecundity["b"],
    "fecundity_Tmax": fecundity["Tmax"],
    "fecundity_c": fecundity["c"],
    "adult_mort_a": adult_mort["a"],
    "adult_mort_b": adult_mort["b"],
    "adult_mort_c": adult_mort["c"],
    "sex_ratio": 0.5,
}

species_dict_clean = to_float(species_dict)
species_config = {"species": species_dict_clean}

with open("species_parameters_option2.yaml", "w") as f:
    yaml.dump(species_config, f, sort_keys=False, indent=2, default_flow_style=False)

print("✅ Option 2 YAML saved as species_parameters_option2.yaml")
```
